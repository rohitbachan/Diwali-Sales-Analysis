Diwali Sales Analysis

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns #import library
```

```python
df=pd.read_csv('Diwali Sales Data.csv',encoding='latin1')
df.head() #top 5 rows
```

```python
#Rename column name
df.rename(columns={'Cust_name':'Customer_name'},inplace=True)
df.head()
```

```python
#No of rows and column
df.shape
```

```python
#Statistics of numeric columns
df.describe().round(2)
```

```python
df.info() #Information of dataset
```

```python
df.drop(['Status','unnamed1'],axis=1,inplace=True) #drop columns
```

```python
df.isnull().sum() #Find null value
```

```python
df['Amount'].mean()
```

```python
df['Amount'].median()
```

```python
df['Amount']=df['Amount'].fillna(df['Amount'].median()) #replacing missing values
```

```python
df['Amount'] = df['Amount'].astype('int') # changing datatype
```

```python
df['Amount'].dtype #check datatype
```

```python
df[['Customer_name','Product_ID','Gender']] 
```

```python
df.duplicated().sum() # sum duplicated
```

```python
df.drop_duplicates() #Removing Duplicates
```

```python
#Total married customer
Total_customer=df['Customer_name'].count()
print('Total customer',Total_customer)
Total_married=df['Marital_Status'].sum()
print('Total married',Total_married)
```

```python
#Average Order Values
Order_Value=df['Orders'].mean()
print(Order_Value)
```

```python
#Product category by sale
Product_category_sale=df.groupby('Product_Category')['Orders'].sum().sort_values(ascending=False).head()
print('Product Wise Sale',Product_category_sale)
plt.bar(Product_category_sale.index,Product_category_sale.values,color='lightgreen')
plt.title('Product category wise Sales')
plt.xlabel('Product Category')
plt.ylabel('Sales')
plt.show()
```

```python
#KPIs(Total orders)
total_order = df['Orders'].sum()
print('Total Order',total_order)
plt.bar(['Total Orders'], [total_order], color='skyblue')
plt.title('Total Orders')
plt.ylabel('Count')
plt.show()
```

```python
# BEST zone sales
TOP_ZONE_SALES=df.groupby('Zone')['Orders'].sum().sort_values(ascending=False).head(5)
print(TOP_ZONE_SALES)
```

```python
#Top Zones
plt.bar(TOP_ZONE_SALES.index,TOP_ZONE_SALES.values,color='Grey')
plt.xlabel('Zones')
plt.show
```

```python
#Top states
TOP_STATE_SALES=df.groupby('State')['Orders'].sum().sort_values(ascending=False).head(5)
print(TOP_STATE_SALES)
```

```python
plt.bar(TOP_STATE_SALES.index,TOP_STATE_SALES.values,color='Yellow')
plt.show
```

```python
#contribution by gender
Gender_by_Order=df.groupby('Gender')['Orders'].sum()
print(Gender_by_Order)
```

```python
Gender_by_Order=df.groupby('Gender')['Orders'].sum()
plt.bar(Gender_by_Order.index,Gender_by_Order.values,color='Green')
plt.show
```

```python
Gender_by_Product=df.groupby('Gender')['Product_Category'].count()
print(Gender_by_Product)
Gender_by_Product.plot(kind='bar', color=['pink','skyblue'])
plt.title('Products Purchased by Gender')
plt.ylabel('Count')
plt.xlabel('Gender')
plt.show()
```

```python
Gender_by_Products=df.groupby('Gender')['Product_Category'].count().sort_values(ascending=False).head(5)
print(Gender_by_Products)
```

# Gender purchased more in each product category

```python
Gender_Product = df.groupby(['Product_Category','Gender'])['Orders'].count().unstack()
print(Gender_Product)
Gender_Product.plot(kind='bar',stacked=True,figsize=(10,6),color=['purple','lightgreen'])
plt.title('Order by Product Category and Gender')
plt.ylabel('count')
plt.show
```

```python
#Top customer
Top_customer=df.groupby('Customer_name')['Orders'].sum().sort_values(ascending=False).head(3)
print(Top_customer)
plt.bar(Top_customer.index,Top_customer.values)
plt.title('Top customers by orders')
plt.xlabel('Top customers')
plt.ylabel('Order')
plt.show
```

```python
occupation_gender_wise=df.groupby(['Occupation','Gender'])['Orders'].count().unstack()
print(occupation_gender_wise)
occupation_gender_wise.plot(kind='bar', stacked=True, figsize=(6,6), color=['purple','skyblue'])
plt.title('Orders by Occupation and Gender')
plt.ylabel('Count')
plt.show()
```

# Conclusion
female customers being the primary drivers of Diwali sales, both in terms of total orders and product variety.
